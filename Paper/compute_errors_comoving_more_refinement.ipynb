```python
import magritte.core     as magritte   # Core functionality
import magritte.plot     as plot       # Plotting
import magritte.tools    as tools      # Save fits
import numpy as np
import os
import time                                            # For timing

from astropy import units              # Unit conversions
```

Define a working directory (you will have to change this). We assume here that the scripts of the this example have already been executed and go back to that working directory.

```python
wdir = "/STER/thomasc/Magritte_comoving_adaptiverays/Magritte-examples/Phantom_3D_more_refinement/"
```

Define file names.

```python
model_file_rec = os.path.join(wdir, 'wind_red_rec.hdf5')   # reduced 3D Phantom Magritte model
model_file_original = os.path.join(wdir, 'model_Phantom_3D_for_comoving.hdf5')
   # original 3D Phantom Magritte model
J_file_feaut = os.path.join(wdir, 'J_feautrier.npy')
J_file_comoving = os.path.join(wdir, 'J_comoving.npy')
J_file_comapp = os.path.join(wdir, 'J_comapp.npy')
J_file_shortchar = os.path.join(wdir, 'J_shortchar.npy')

J_file_original_comoving = os.path.join(wdir, 'J_original_comoving.npy')
J_file_original_feautrier = os.path.join(wdir, 'J_original_feautrier.npy')
J_file_original_comapp = os.path.join(wdir, 'J_original_comapp.npy')

NLTE_J_file_feaut = os.path.join(wdir, 'NLTE_J_rec.npy')
NLTE_J_file_comoving = os.path.join(wdir, 'NLTE_J_comoving.npy')
NLTE_J_file_comoving_approx = os.path.join(wdir, 'NLTE_J_comoving_approx.npy')

# model_file_rec_feautrier_save = os.path.join(wdir, 'wind_rec_feautrier_save.hdf5')   # reduced 3D Phantom Magritte model
# model_file_rec_comoving_save = os.path.join(wdir, 'wind_rec_comoving_save.hdf5')   # reduced 3D Phantom Magritte model
# model_file_rec_comapp_save = os.path.join(wdir, 'wind_rec_comapp_save.hdf5')   # reduced 3D Phantom Magritte model

model_file_original_feautrier_save = os.path.join(wdir, 'model_Phantom_3D_feautrier_save.hdf5')   # original 3D Phantom Magritte model
model_file_original_comoving_save = os.path.join(wdir, 'model_Phantom_3D_comoving_save.hdf5')   # original 3D Phantom Magritte model
model_file_original_comapp_save = os.path.join(wdir, 'model_Phantom_3D_comapp_save.hdf5')   # original 3D Phantom Magritte model
```

Load all models

```python
model_original = magritte.Model(model_file_original)
model_rec = magritte.Model(model_file_rec)
```

```python
model_rec.compute_spectral_discretisation ()
model_rec.compute_inverse_line_widths     ()
model_rec.compute_LTE_level_populations   ()

model_original.compute_spectral_discretisation ()
model_original.compute_inverse_line_widths     ()
model_original.compute_LTE_level_populations   ()
```

Initialize all models.

## Computing mean line intensities

Use sparse feautrier solver to compute radiation field

```python
start = time.time()
model_rec.compute_radiation_field_feautrier_order_2_sparse()
end = time.time()
print("Feautrier computation time: ", end - start)
```

Save in .npy file.

```python
np.save(J_file_feaut, np.array(model_rec.lines.lineProducingSpecies[0].J))
```

```python
start = time.time()
model_rec.compute_radiation_field_comoving()
end = time.time()
print("Comoving computation time: ", end - start)
np.save(J_file_comoving, np.array(model_rec.lines.lineProducingSpecies[0].J))
```

```python
start = time.time()
model_rec.compute_radiation_field_comoving_local_approx()
end = time.time()
print("Comoving local approx computation time: ", end - start)
np.save(J_file_comapp, np.array(model_rec.lines.lineProducingSpecies[0].J))
```

## NLTE computations

The goal of the new type of solvers is to compute NLTE a bit faster, thus we will compute NLTE on the original grid; and compare the speedups. Note: we will not be using Ng-acceleration, as this will complicate the analysis a lot...

```python
model_original.compute_LTE_level_populations()
start = time.time()
model_original.compute_level_populations_comoving(False, 5)#Without ng-acceleration, because it becomes a bit cumbersome to compare if I also include it
end = time.time()
print("NLTE Comoving computation time on reduced: ", end - start)
np.save(NLTE_J_file_comoving, np.array(model_original.lines.lineProducingSpecies[0].J))
model_original.write(model_file_original_comoving_save)
```

Also compare with comoving approx version

```python
model_original.compute_LTE_level_populations   ()
start = time.time()
model_original.compute_level_populations_comoving_local_approx(False, 5)#Without ng-acceleration, because it becomes a bit cumbersome to compare if I also include it
end = time.time()
print("NLTE Comoving computation time on reduced: ", end - start)
np.save(NLTE_J_file_comoving_approx, np.array(model_original.lines.lineProducingSpecies[0].J))
model_original.write(model_file_original_comapp_save)
```

And evidently versus the baseline feautrier solver.

```python
model_original.compute_LTE_level_populations   ()
start = time.time()
model_original.compute_level_populations_sparse(False, 5)
end = time.time()
print("NLTE feautrier computation time on reduced: ", end - start)
np.save(NLTE_J_file_feaut, np.array(model_original.lines.lineProducingSpecies[0].J))
model_original.write(model_file_original_feautrier_save)
```
